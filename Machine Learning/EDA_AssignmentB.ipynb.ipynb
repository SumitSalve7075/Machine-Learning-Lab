```python
# Remove S.No. column from data
print(data.columns)
```

```python
data = data.drop(columns=['PassengerId'])
```

```python
data.info()
```

```python
# Recreate PassengerId (1 to 891)
data['PassengerId'] = range(1, len(data) + 1)

# Move PassengerId to the first column
cols = data.columns.tolist()
cols = ['PassengerId'] + [col for col in cols if col != 'PassengerId']
data = data[cols]
```

```python
data.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
# Bar chart: Survived vs Sex
sns.countplot(x='Sex', hue='Survived', data=data)
plt.title('Survival Count by Gender')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

```python
# Pie chart: Overall survival
survived_counts = data['Survived'].value_counts()
labels = ['Not Survived', 'Survived']
plt.pie(survived_counts, labels=labels, autopct='%1.1f%%', colors=['#ff9999','#66b3ff'])
plt.title('Survival Distribution')
plt.axis('equal')
plt.show()
```

```python
# Histogram: Age distribution
plt.hist(data['Age'].dropna(), bins=20, color='skyblue', edgecolor='black')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

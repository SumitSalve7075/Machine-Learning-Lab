```python
import matplotlib.pyplot as plt
df['Survived'].value_counts().plot(kind='bar', color=['red', 'green'])
plt.title('Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.xticks([0, 1], ['Did Not Survive', 'Survived'])
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Titanic-Dataset.csv")

# Create a grouped dataframe: count of survivors by 'Sex' and 'Pclass'
grouped = df.groupby(['Pclass', 'Sex'])['Survived'].sum().unstack()

# Plot the stacked bar chart
grouped.plot(kind='bar', stacked=True, color=['skyblue', 'lightpink'], figsize=(8, 6))

plt.title('Survival Count by Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Survivors')
plt.legend(title='Gender')
plt.xticks(rotation=0)
plt.grid(axis='y')
plt.show()
```
